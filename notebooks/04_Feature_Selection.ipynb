# Phase 4 : Feature Selection using mRMR

## Objectives

- Load cleaned dataset
- Encode target labels
- Apply mRMR
- Select Top Features
- Save selected dataset

```python
# ============================================================
# Import Libraries
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries Imported")
```

```python
PROJECT_DIR = Path(r"A:\Network-Intrusion-Detection-System")

DATASET_PATH = PROJECT_DIR/"dataset"/"cleaned"/"cleaned_cicids2017.csv"

df = pd.read_csv(DATASET_PATH)

print(df.shape)
```

```python
X = df.drop(columns=["Label"])

y = df["Label"]

print(X.shape)

print(y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

print("Classes")

print(encoder.classes_)
```

```python
!pip install mrmr-selection
```

```python
from mrmr import mrmr_classif
```

```python
selected_features = mrmr_classif(
    X=X,
    y=pd.Series(y_encoded),
    K=30
)
```

```python
print(selected_features)
```

```python
selected_df = X[selected_features]

selected_df["Label"] = y.values

selected_df.head()
```

```python
OUTPUT_PATH = PROJECT_DIR/"dataset"/"cleaned"/"selected_features.csv"

selected_df.to_csv(OUTPUT_PATH,index=False)

print("Saved Successfully")

print(OUTPUT_PATH)
```
