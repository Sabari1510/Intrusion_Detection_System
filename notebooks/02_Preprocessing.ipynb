```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries Imported Successfully")
```

# Phase 2 : Data Preprocessing

## Objectives

- Load merged dataset
- Remove duplicate records
- Handle infinite values
- Handle missing values
- Clean column names
- Save cleaned dataset

```python
# ==========================================================
# Project Paths
# ==========================================================

PROJECT_DIR = Path(r"A:\Network-Intrusion-Detection-System")

DATASET_PATH = PROJECT_DIR / "dataset" / "cleaned" / "merged_dataset.csv"

print(DATASET_PATH)
```

```python
# ==========================================================
# Load Merged Dataset
# ==========================================================

merged_df = pd.read_csv(DATASET_PATH)

print("Dataset Loaded Successfully")

print(merged_df.shape)
```

```python
merged_df.columns = merged_df.columns.str.strip()

print("Column Names Cleaned")
```

```python
merged_df.columns
```

```python
before_rows = merged_df.shape[0]

merged_df = merged_df.drop_duplicates()

after_rows = merged_df.shape[0]

print(f"Rows Before : {before_rows}")

print(f"Rows After : {after_rows}")

print(f"Duplicates Removed : {before_rows-after_rows}")
```

```python
merged_df.replace([np.inf,-np.inf],np.nan,inplace=True)

print("Infinite values converted to NaN")
```

```python
merged_df.fillna(0,inplace=True)

print("Remaining Missing Values")

print(merged_df.isnull().sum().sum())
```

```python
print("Dataset Shape")

print(merged_df.shape)

print()

print("Missing Values")

print(merged_df.isnull().sum().sum())

print()

print("Infinite Values")

print(np.isinf(
    merged_df.select_dtypes(include=[np.number])
).sum().sum())
```

```python
OUTPUT_PATH = PROJECT_DIR / "dataset" / "cleaned" / "cleaned_cicids2017.csv"

merged_df.to_csv(OUTPUT_PATH,index=False)

print("Dataset Saved Successfully")

print(OUTPUT_PATH)
```
